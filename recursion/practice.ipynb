```python
count = 0

def func():
    global count

    count += 1

    if count > 10: return

    print("something + " + str(count))

    func()

def main():
    func()

main()
```

```python
def func(x):
    if x == 3: return

    print("something + " + str(x))
    func(x+1)

def main():
    func(0)

main()
```

```python
# print name n times
count = 0

def printer():
    global count
    if count == 5:
        return

    count += 1
    print("davis")
    printer()

def main():
    printer()

main()
```
